```python
import pandas as pd
from sklearn.metrics import classification_report

df = pd.read_csv('ml_features_and_labels.csv')
```

```python
# ===========================
# Features used for detection
# ===========================
rule_features = [
    "e3_cert_parsed",
    "e1_alg_suite_Unknown(0x11eb)",
    "e1_alg_suite_Unknown(0x11ec)",
    "e1_alg_suite_nan",
    "e3_sig_alg_Unknown(0x0028)",
    "e9_conn_outcome_Failure",
    "e9_conn_outcome_Incomplete",
    "e8_proto_context_TLSv1.2_or_older",
    "e1b_tls_version_SSL3.0",
    "e1b_ciphersuite_2",
    "e1b_ciphersuite_54"
]

# ===========================
# Rule-Based Detection Engine
# ===========================
def detect(row):

    if row["e3_cert_parsed"] == True:
        return 1

    if row["e1_alg_suite_Unknown(0x11eb)"] == True:
        return 1

    if row["e1_alg_suite_Unknown(0x11ec)"] == True:
        return 1

    if row["e1_alg_suite_nan"] == True:
        return 1

    if row["e3_sig_alg_Unknown(0x0028)"] == True:
        return 1

    if row["e9_conn_outcome_Failure"] == True:
        return 1

    if row["e9_conn_outcome_Incomplete"] == True:
        return 1

    if row["e8_proto_context_TLSv1.2_or_older"] == True:
        return 1

    if row["e1b_tls_version_SSL3.0"] == True:
        return 1

    if row["e1b_ciphersuite_2"] == True:
        return 1

    if row["e1b_ciphersuite_54"] == True:
        return 1
    if row["e1_alg_suite_mlkem1024"]:
        return 0

    return None


# Apply rules
df["Rule_Label"] = df.apply(detect, axis=1)
```
