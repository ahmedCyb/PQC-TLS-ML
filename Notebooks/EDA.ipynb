### Missing Values

```python
print("="*60)
print("Missing Values")
print("="*60)

missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

display(pd.DataFrame({
    "Missing Values": missing,
    "Percentage": (missing / len(df) * 100).round(2)
}))
```

### Checking for Possible Leakage

```python
exclude_cols = ["label", "ID", "split", "taxonomy"]
bool_cols = [col for col in df.columns if df[col].dtype == "bool"]

suspect_cols = [col for col in df.columns if col not in exclude_cols and col in bool_cols]
    
for col in suspect_cols:
    ct = pd.crosstab(df[col], df["label"], normalize="index") * 100
    print(ct.round(2))
```
